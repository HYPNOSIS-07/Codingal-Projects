```python
print("\nWelcome To SpaceX Mars Civilisation Eligibility Checker")
age=int(input("\nPlease Enter Your Age In Years:   "))
if age >=18:
  coding=input("\nDo U have coding/technical skills? <Yes/No>:  ").lower()
  health=input("\nAre You sure THAT YOU  ARE FIT for training and settlement ?<Yes/No:   ").lower()
  if coding == "yes" and health =="yes":
    print("\nCongrats !! You Are Eligible To Fill The Form !!!")
  elif coding =="no" and health=="yes":
    print("\nYou can take our coding classes for the training of your skills!! ")
  else:
    print("\nYou are not eligible to enroll in this program!! Try Next Time!!")
else:
  print("\nYou are Underage!! Try When You Are 18+")
```
